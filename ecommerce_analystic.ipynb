```python
sales.duplicated("customer_id").sum()

sales= sales.sort_values("order_date")

first_purchase= (sales.groupby("customer_id")["order_date"].min())

new_customers_2026 = first_purchase[
    (first_purchase >= "1/1/2026") &
    (first_purchase <= "1/12/2026")
]

print(new_customers_2026)

# calculate the new customer form the data sets
```

```python

sales["month_number"] = sales["order_date"].dt.month

sales_2026= sales[sales["order_date"].dt.year== 2026]

monthly_revenue=  (
    sales_2026.groupby(["month_number","month"])["revenue"]
    .sum()
    .sort_index())

print(monthly_revenue)


# max sales in year 2026
```

```python

monthly_revenue.plot(kind="line",marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
category_revenue= (sales.groupby("product_category")["revenue"]
            .sum()
            .sort_index()
    )

print(category_revenue)
```

```python
plt.figure(figsize=(12, 6))

ax = category_revenue.plot(kind="bar")

plt.title("Category Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")

ax.bar_label(ax.containers[0], fmt='%.0f', padding=3)
from matplotlib.ticker import FuncFormatter

plt.gca().yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f'{x/1000:.0f}k')
)


plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
regional_revenue= (
    sales.groupby("region")["revenue"]
    .sum()
    .sort_values(ascending= False)
)

print(regional_revenue)
```

```python
regional_revenue.plot(kind="bar")

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=0)

from matplotlib.ticker import FuncFormatter

plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"{x/1000:.0f}K"))
plt.show()
```

```python
# Filter dataframe for 5-star ratings only

five_star_sales = sales[sales['customer_rating'] == 5]

plt.figure(figsize=(10, 5))
ax = sns.countplot(
    data=five_star_sales,
    x='delivery_days'    
)

# Display exact counts on top of each bar
ax.bar_label(ax.containers[0])

plt.title('Total 5-Star Ratings by Delivery Days')
plt.xlabel('Delivery Days')
plt.ylabel('Count of 5-Star Ratings')
plt.show()
```
